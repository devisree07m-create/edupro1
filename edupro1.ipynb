```python
import pandas as pd

# Load the uploaded CSV file
file_path = 'EduPro Online Platform.xlsx - Users.csv'
df = pd.read_csv(file_path)

# Display basic information and the first few rows
print(df.info())
display(df.head())
```

```python
# Detailed Descriptive Statistics for the Dataset
print("=== Summary Statistics for Age ===")
print(df['Age'].describe())

print("\n=== Gender Distribution ===")
print(df['Gender'].value_counts())

print("\n=== Check for Null/Missing Values ===")
print(df.isnull().sum())

print("\n=== Unique Users Count ===")
print(f"Total Rows: {len(df)}")
print(f"Unique UserIDs: {df['UserID'].nunique()}")
print(f"Unique UserNames: {df['UserName'].nunique()}")
print(f"Unique Emails: {df['Email'].nunique()}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style of the plots
sns.set_theme(style="whitegrid")

# Create the bar chart for Gender distribution
plt.figure(figsize=(6, 4))
gender_counts = df['Gender'].value_counts()
sns.barplot(x=gender_counts.index, y=gender_counts.values, palette='pastel', hue=gender_counts.index, legend=False)

# Add title and labels
plt.title('Gender Distribution on EduPro Platform', fontsize=14, pad=15)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Users', fontsize=12)

# Display data labels on top of the bars
for index, value in enumerate(gender_counts.values):
    plt.text(index, value + 20, str(value), ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()
```
